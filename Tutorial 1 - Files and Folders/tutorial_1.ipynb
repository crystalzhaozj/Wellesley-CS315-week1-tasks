# 3. Getting File Attributes

```python
with os.scandir(basepath) as dir_contents:
    for entry in dir_contents:
        info = entry.stat() # this is the code to get attributes
        print(info.st_mtime) # the time the content of the file was last modified
```

```python
from datetime import datetime
from os import scandir

def convert_date(timestamp):
    d = datetime.utcfromtimestamp(timestamp)
    formated_date = d.strftime('%d %b %Y')
    return formated_date

def get_files():
    dir_entries = scandir(basepath)
    for entry in dir_entries:
        if entry.is_file():
            info = entry.stat()
            print(f'{entry.name}\t Last Modified: {convert_date(info.st_mtime)}')
```

```python
get_files()
```

# 5. Filenames

```python
import fnmatch

for file_name in os.listdir(basepath):
    if fnmatch.fnmatch(file_name, '*.txt'):
        print(file_name) # anything that has a txt extension
```

recursive searching with Glob

```python
import glob

os.chdir('/Users/crystalzhao/Desktop/~WELLESLEY~/FALL_26/CS 315/Wellesley-CS315-week1-tasks')
print(os.getcwd())

for file in glob.iglob('**/*.ipynb', recursive=True):
    print(file) # this only searches in current directory though recursively 
```

# 9. Archives

Creating a ZIP file:

```python
import zipfile

file_list = ['copy.txt', 'example/']

with zipfile.ZipFile('new.zip', 'w') as new_zip:
    for name in file_list:
        new_zip.write(name)
```

```python
with zipfile.ZipFile('new.zip', 'a') as new_zip:
    new_zip.write('copy.txt')
    new_zip.write('tutorial_1.ipynb')
```

TAR archives: uncompressed file archives like ZIP.
