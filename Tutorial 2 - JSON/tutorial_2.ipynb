## 1. Writing JSON with Python

Serialization: converting data into the JSOn format - transforming data into a series of bytes for storage or transmission over a network

Deserialization: decoding data from JSON back into usable form within Python

json.dumps: convert Python dictionary into a JSON-formatted-string

```python
import json

food_ratings = {"organic dog food": 2, "human food": 10}
json.dumps(food_ratings)
```

You do NOT create a JSON data type - it's just a Python string. 

For non-strings like booleans, they become JSON booleans

```python
numbers_present = {1: True, 2: True, 3: False}
json.dumps(numbers_present)
```

Keys must be str, int, float, bool, or None - NOT a tuple
